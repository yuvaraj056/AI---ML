```python
# Understanding Dendrograms in Hierarchical Clustering

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import dendrogram, linkage

# Step 1: Create sample data
X, y = make_blobs(n_samples=30, centers=3, random_state=42)


# Step 2: Generate linkage matrix using Ward's method
linked = linkage(X, method='ward')

# Step 3: Plot the dendrogram
plt.figure(figsize=(10, 6))
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title("Dendrogram - Visualizing the Clustering Process")
plt.xlabel("Sample Index")
plt.ylabel("Distance")
plt.grid(True)
plt.tight_layout()
plt.show()
```
