```python
# Question 3: Implement t-SNE on a Dataset
# Description: Use t-SNE to visualize high-dimensional data.
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE

# Load and standardize data
data = load_iris()
X = data.data
y = data.target
X_scaled = StandardScaler().fit_transform(X)

# Apply t-SNE to reduce dimensions to 2
tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

# Plot the t-SNE results
plt.figure(figsize=(8,6))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', edgecolor='k', s=70)
plt.xlabel('t-SNE Dimension 1')
plt.ylabel('t-SNE Dimension 2')
plt.title('t-SNE Visualization of Iris Dataset')
handles, _ = scatter.legend_elements()
plt.legend(handles, data.target_names)
plt.grid(True)
plt.show()
```
