```python
# Question 5: Visual Comparison of PCA and t-SNE
# Description: Compare PCA and t-SNE on the same dataset visually.
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

# Load and scale the Iris dataset
data = load_iris()
X = StandardScaler().fit_transform(data.data)
y = data.target
# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Apply t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
X_tsne = tsne.fit_transform(X)

# Plot side-by-side comparison
fig, axs = plt.subplots(1, 2, figsize=(14, 6))

# PCA Plot
scatter1 = axs[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', edgecolor='k', s=60)
axs[0].set_title("PCA Visualization")
axs[0].set_xlabel("PC 1")
axs[0].set_ylabel("PC 2")

# t-SNE Plot
scatter2 = axs[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', edgecolor='k', s=60)
axs[1].set_title("t-SNE Visualization")
axs[1].set_xlabel("Dim 1")
axs[1].set_ylabel("Dim 2")

# Add legends
for ax in axs:
    handles, _ = scatter1.legend_elements()
    ax.legend(handles, data.target_names, loc="best")

plt.tight_layout()
plt.show()
```
