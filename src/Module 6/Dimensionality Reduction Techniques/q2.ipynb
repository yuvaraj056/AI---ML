```python
# Question 2: Geometry of PCA
# Description: Compute PCA on a simple 2D matrix and manually derive its principal components.
```

```python
import numpy as np

# Step 1: Define a simple 2D dataset (each row is a data point)
X = np.array([[2.5, 2.4],
              [0.5, 0.7],
              [2.2, 2.9],
              [1.9, 2.2],
              [3.1, 3.0],
              [2.3, 2.7],
              [2.0, 1.6],
              [1.0, 1.1],
              [1.5, 1.6],
              [1.1, 0.9]])

# Step 2: Compute the mean of each feature and center the data
mean_vec = np.mean(X, axis=0)
X_centered = X - mean_vec

# Step 3: Calculate the covariance matrix
cov_matrix = np.cov(X_centered, rowvar=False)

# Step 4: Compute eigenvalues and eigenvectors of covariance matrix
eig_vals, eig_vecs = np.linalg.eigh(cov_matrix)  # eigh for symmetric matrices

# Step 5: Sort eigenvalues and eigenvectors in descending order
sorted_indices = np.argsort(eig_vals)[::-1]
eig_vals = eig_vals[sorted_indices]
eig_vecs = eig_vecs[:, sorted_indices]

# Step 6: Display results
print("Mean Vector:\n", mean_vec)
print("\nCovariance Matrix:\n", cov_matrix)
print("\nEigenvalues:\n", eig_vals)
print("\nEigenvectors:\n", eig_vecs)

# Step 7: Project data onto principal components
X_pca = X_centered @ eig_vecs

print("\nData projected onto principal components:\n", X_pca)
```
