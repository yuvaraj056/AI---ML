```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_digits  # substitute for CIFAR-10
import matplotlib.pyplot as plt

# Load a smaller image dataset (e.g., handwritten digits)
digits = load_digits()
X = digits.data
y = digits.target

# Scale the data
X_scaled = StandardScaler().fit_transform(X)

# Reduce to 2 dimensions using PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Plot the 2D PCA result
plt.figure(figsize=(8,6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='tab10', edgecolor='k', s=50)
handles, _ = scatter.legend_elements()
plt.legend(handles, [str(i) for i in range(10)], title="Digit")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("PCA on Handwritten Digits Dataset")
plt.grid(True)
plt.show()
```
