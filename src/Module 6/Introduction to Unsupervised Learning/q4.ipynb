```python
# Question 3: Basic Association Rule Learning using Apriori
# Description: Use the Apriori algorithm to find association rules in a sample transaction dataset.
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from sklearn.preprocessing import MultiLabelBinarizer

# Sample transaction dataset (list of items bought in each transaction)
transactions = [
    ['milk', 'bread', 'butter'],
    ['beer', 'bread'],
    ['milk', 'bread', 'beer', 'butter'],
    ['bread', 'butter'],
    ['milk', 'beer'],
    ['milk', 'bread', 'butter', 'beer'],
]

# Step 1: Convert the list of transactions into a one-hot encoded DataFrame
mlb = MultiLabelBinarizer()
onehot = pd.DataFrame(mlb.fit_transform(transactions), columns=mlb.classes_)

# Step 2: Use apriori algorithm to find frequent itemsets with a minimum support threshold
frequent_itemsets = apriori(onehot, min_support=0.3, use_colnames=True)

# Step 3: Generate association rules from frequent itemsets
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

# Display the rules
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
