```python
# Question 9: Hierarchical Clustering for Customer Segmentation
# Description: Use hierarchical clustering to segment customers and visualize the dendrogram.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler


# Sample customer data: [Annual Income (k$), Spending Score (1-100)]
data = np.array([
    [15, 39],
    [16, 81],
    [17, 6],
    [18, 77],
    [19, 40],
    [20, 76],
    [21, 6],
    [22, 94],
    [23, 3],
    [24, 72]
])

# Convert to DataFrame for easier handling
df = pd.DataFrame(data, columns=['Annual Income (k$)', 'Spending Score (1-100)'])

# Step 1: Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Step 2: Compute linkage matrix using 'ward' method (minimizes variance within clusters)
Z = linkage(X_scaled, method='ward')

# Step 3: Plot dendrogram to visualize the clusters
plt.figure(figsize=(10, 6))
dendrogram(Z, labels=df.index, leaf_rotation=45, leaf_font_size=12)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Customer Index')
plt.ylabel('Distance')
plt.show()
```
