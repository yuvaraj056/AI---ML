```python
# Question 6 : Comparing K-Means and Hierarchical Clustering
# Description: Compare the results of K-Means and Hierarchical Clustering on a dataset and discuss the potential differences in their clustering patterns.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import linkage, fcluster
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.7, random_state=42)

plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Original Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X)
Z = linkage(X, method='ward')
hier_labels = fcluster(Z, t=3, criterion='maxclust')
fig, axs = plt.subplots(1, 2, figsize=(14, 6))

# K-Means result
axs[0].scatter(X[:, 0], X[:, 1], c=kmeans_labels, cmap='viridis', s=50)
axs[0].set_title("K-Means Clustering")
axs[0].set_xlabel("Feature 1")
axs[0].set_ylabel("Feature 2")

# Hierarchical result
axs[1].scatter(X[:, 0], X[:, 1], c=hier_labels, cmap='viridis', s=50)
axs[1].set_title("Hierarchical Clustering")
axs[1].set_xlabel("Feature 1")
axs[1].set_ylabel("Feature 2")
plt.show()
```
