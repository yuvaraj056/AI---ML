```python
# Question 3: Implementing DBSCAN Clustering
# Description: Perform DBSCAN clustering on a dataset with noise and visualize the clusters.
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
# Create a dataset with noise and non-linear clusters
X, y = make_moons(n_samples=300, noise=0.1, random_state=0)

# Visualize raw data
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Noisy Dataset (make_moons)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
# Apply DBSCAN
db = DBSCAN(eps=0.2, min_samples=5)
y_db = db.fit_predict(X)
# Plotting results
plt.figure(figsize=(8, 6))

# Identify core and noise points
core_samples_mask = (y_db != -1)
unique_labels = set(y_db)

# Define colors for clusters and noise
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_labels)))

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = 'k'
    class_member_mask = (y_db == k)
    plt.plot(X[class_member_mask, 0], X[class_member_mask, 1], 'o', markerfacecolor=col, markersize=6)

plt.title('DBSCAN Clustering Result')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
