Title: Grouping & Aggregating Data using Pandas<br>
Objective: Learn how to group data and perform aggregations on these groups.

Task 1: Grouping by a Single Column<br>

Task: Group the dataset by 'region' and calculate total sales per region.<br>
Steps:<br>
10. Load the dataset.<br>
11. Use groupby('region') on the DataFrame.<br>
12. Apply .sum() to the 'sales' column.

```python
import pandas as pd
import os

# Step 0: Create a sample CSV file with 'region' and 'sales' columns
sample_data = {
    'region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'sales': [250, 150, 300, 200, 100, 180, 220, 160]
}

df_sample = pd.DataFrame(sample_data)
csv_filename = 'region_sales_data.csv'
df_sample.to_csv(csv_filename, index=False)
print(f"Sample CSV file '{csv_filename}' created.\n")

# Check available CSV files
print("Available CSV files:")
for file in os.listdir():
    if file.endswith('.csv'):
        print(f"- {file}")

# Load the created CSV file (the one we just saved)
df = pd.read_csv(csv_filename)

# Group by 'region' and sum 'sales'
region_sales = df.groupby('region')['sales'].sum()

print("\nTotal sales per region:")
print(region_sales)
```

Task 2: Grouping by Multiple Columns<br>

Task: Group the dataset by 'region' and 'category', then find the average sales.<br>
Steps:<br>
13. Group by ['region', 'category'].<br>
14. Use .mean() on the 'sales' column.<br>
15. Examine the resulting DataFrame structure.

```python
import pandas as pd
import os

# Step 1: Create sample data with 'region' and 'sales'
data = {
    'region': ['North', 'South', 'East', 'West', 'North', 'South'],
    'sales': [250, 150, 300, 200, 100, 180]
}

# Step 2: Create DataFrame and save to CSV
df = pd.DataFrame(data)
csv_filename = 'region_sales.csv'
df.to_csv(csv_filename, index=False)
print(f"CSV file '{csv_filename}' created.\n")

# Step 3: Load CSV file
df_loaded = pd.read_csv(csv_filename)
print("Loaded CSV data:")
print(df_loaded)

# Step 4: Group by 'region' and sum 'sales'
region_sales = df_loaded.groupby('region')['sales'].sum()
print("\nTotal sales per region:")
print(region_sales)
```

Task 3: Aggregating Multiple Functions<br>

Task: Group data by 'category' and apply multiple aggregation functions (sum and count) on 'quantity'.<br>
Steps:<br>
16. Group by 'category'.<br>
17. Use .agg(['sum', 'count']) on 'quantity'.<br>
18. Analyze the result to understand how multiple aggregations work.

```python
import pandas as pd
import os

# Step 0: Create a sample CSV file with 'category' and 'quantity' columns
sample_data = {
    'category': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'C', 'A'],
    'quantity': [10, 15, 5, 7, 20, 10, 12, 8, 9, 6]
}

df_sample = pd.DataFrame(sample_data)
csv_filename = 'category_quantity_data.csv'
df_sample.to_csv(csv_filename, index=False)
print(f"Sample CSV file '{csv_filename}' created.\n")

# Now confirm the CSV file is present
print("Available files after creation:")
for file in os.listdir():
    if file.endswith('.csv'):
        print(f"- {file}")

# Step 16: Load the CSV file
df = pd.read_csv(csv_filename)

# Step 17: Group by 'category' and apply sum and count aggregations on 'quantity'
category_agg = df.groupby('category')['quantity'].agg(['sum', 'count'])

# Step 18: Analyze the result
print("\nMultiple aggregations (sum and count) by category:")
print(category_agg)

print("\nAnalysis:")
print(f"DataFrame shape: {category_agg.shape}")
print(f"Columns: {category_agg.columns.tolist()}")
print(f"Index name: {category_agg.index.name}")

# Calculate average quantity per category
category_agg['average'] = category_agg['sum'] / category_agg['count']
print("\nWith calculated average quantity:")
print(category_agg)

# Sort by total quantity descending
print("\nCategories sorted by total quantity (descending):")
print(category_agg.sort_values('sum', ascending=False))
```
